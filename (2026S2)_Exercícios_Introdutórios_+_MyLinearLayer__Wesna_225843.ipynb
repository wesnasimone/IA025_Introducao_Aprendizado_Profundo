## Exercício 1.5
Dicionários são uma estrutura extremamente poderosa que associa um valor a outro valor, com tempo de acesso O(1).

Crie um dicionário com os `k` itens mais frequentes de uma lista.

Por exemplo, dada a lista de itens `L=['a', 'a', 'd', 'b', 'd', 'c', 'e', 'a', 'b', 'e', 'e', 'a']` e `k=2`, o resultado deve ser um dicionário cuja chave é o item e o valor é a sua frequência: {'a': 4, 'e': 3}

Otimizações são permitidas e encorajadas, limitadas a uso de funções de bibliotecas nativas do python.

````python
from collections import Counter

def top_k(L, k):
    # TODO Escreva aqui o código
    a = Counter(L) # cria um dicionario com a quantidade de cada elemento
    b = a.most_common(k) # pega os elementos mais frequentes de acordo com k
    return dict(b)
````

````python
L = ['f', 'a', 'a', 'd', 'b', 'd', 'c', 'e', 'a', 'b', 'e', 'e', 'a', 'd']
k = 3
resultado = top_k(L=L, k=k)
print(f'resultado: {resultado}')
````

## Exercício 1.6

Em processamento de linguagem natural, é comum convertemos as palavras de um texto para uma lista de identificadores dessas palavras, processo chamado de tokenização. Dado o dicionário `V` abaixo onde as chaves são palavras e os valores são seus respectivos identificadores, converta o texto `T` para uma lista de identificadores.

Palavras que não existem no dicionário V deverão ser convertidas para o identificador do token `unknown`.

O código deve ser insensível a maiúsculas (case-insensitive).

Se atente que pontuações (vírgulas, ponto final, etc) também são consideradas palavras.

Dica: use regex, é possível encontrar facilmente implementações para a tarefa acima na internet somente com bibliotecas do python.

````python
import re

def tokens_to_ids(text, vocabulary):
    # TODO: escreva o código aqui.
    #passo 1: quebrar as palavras
    text_modificado = text.lower()   #transforma texto em minusculas
    regex = r"\w+|[^\w\s]"           #pega todas as palavras mas desconsidera espaços ou numeros (isso faz com que pontuação seja identificada)
    resultado = re.findall(regex, text_modificado)

    #passo 2: atribuir identificador
    id = []
    for chave in resultado:
      if chave in vocabulary:
        id.append(vocabulary[chave])
      else:
        id.append(-1)
    return id
````

Mostre que sua implementação esta correta com um exemplo pequeno:

````python
V = {'eu': 1, 'de': 2, 'gosto': 3, 'comer': 4, '.': 5, 'unknown': -1}
T = 'Eu gosto de comer pizza.'

print(tokens_to_ids(T, V))
````

Mostre que sua implementação é eficiente com um exemplo grande:

````python
V = {'eu': 1, 'de': 2, 'gosto': 3, 'comer': 4, '.': 5, 'unknown': -1}
T = ' '.join(1_000_000 * ['Eu gosto de comer pizza.'])
````

````python
%%timeit
resultado = tokens_to_ids(T, V)
````

## Exercício 1.7

Em aprendizado profundo é comum termos que lidar com arquivos muito grandes.

Dado um arquivo de texto onde cada item é separado por `\n`, escreva um programa que amostre `k` itens desse arquivo aleatoriamente.

Nota 1: Assuma amostragem de uma distribuição uniforme, ou seja, todos os itens tem a mesma probablidade de amostragem.

Nota 2: Assuma que o arquivo não cabe em memória.

Nota 3: Utilize apenas bibliotecas nativas do python.

````python
import linecache

def sample(path: str, k: int, qtd_linhas: int):
    # TODO Escreva o seu código aqui
    lista = []
    #foi inserido mais uma variavel qtd_delinhas para sortear aleatoriamente as linhas a serem lidas
    linhas = random.sample(range(0, qtd_linhas-1), k)
    for c in linhas:
      linha = linecache.getline(path, c) #função mais eficiente para ler linhas especificas usando a memória cache
      lista.append(linha)
    return lista
````

Mostre que sua implementação está correta com um exemplo pequeno:

````python
# Esse código gera um arquivo de exemplo
filename = 'small.txt'
total_size = 100
n_samples = 10

with open(filename, 'w') as fout:
    fout.write('\n'.join(f'line {i}' for i in range(total_size)))
#

samples = sample(path=filename, k=n_samples, qtd_linhas=total_size)
print(samples)
print(len(samples) == n_samples)
````

| Decimal | Binary | One-hot

Implemente a função one_hot(y, n_classes) que codifique o vetor de inteiros y que possuem valores entre 0 e n_classes-1.


````python
def one_hot(y, n_classes):
    # Escreva seu código aqui.
    code = []
    for pos in y:
      vetor = np.zeros(n_classes)
      vetor[pos]=1
      code.append(vetor)
    return code
````

Conceitos de orientação a objetos são muito utilizados em programação PyTorch e Deep Learning. Transformações e datasets são comumente mapeados para classes com definição de \_\_init\_\_ (construtor) para operações de inicialização (o que acontece quando a sintaxe `objeto = Classe(args)` é usada).

## Exercício 3.1

Aproveite seu código de normalização de vetores, e crie uma classe que:

1) Normaliza um array de qualquer dimensão para valores entre 0 e 1 usando máximo e mínimo.

$$
C = \frac{A - A_{\min}}{A_{\max} - A_{\min}}
$$

2) Normaliza um array de qualquer dimensão para que tenha a mesma média e desvio padrão que outro array de referência (referência deve ser passada na inicialização da classe). Pesquise como fazer isso.


````python
# TODO Escreva seu código do Normalizer aqui.
class Normalizer():
    def __init__(self, ref=None):
      self.ref = ref

    def min_max(self, A):
      self.A = np.array(A)

      #passo 1: determinar menor e maior valor da matriz A
      minimo = self.A.min()
      maximo = self.A.max()

      #passo 2: normalizar
      C = (self.A - minimo)/(maximo - minimo)

      return C

    def __call__(self, A):
      self.A = np.array(A)

      media = np.mean(self.ref)
      desvio_padrao = np.std(self.ref)

      # Normaliza
      C = (self.A - media) / desvio_padrao

      return C
````

````python
array_a = [-1, 1, 0]
array_b = [1.4, 0.8, 0.3, 2.5]

# Inicialização da classe Normalizer sem passar nenhum array de referência
normalizer = Normalizer()
````

````python
# Normalizar baseado em min-max
min_max_array = normalizer.min_max(array_a)
min_max_array
````

````python
# Teste da inicialização da classe Normalizer, agora usando array_b como referência de média e desvio padrão.
normalizer = Normalizer(array_b)

# Normalizar array a baseado em stats do array_b
normalized_array = normalizer(array_a)
print(normalized_array)
````

Termine a implementação de `ImageDataset`, que recebe o dataset base (ex.: MNIST) e a flag `normalize`. Quando `normalize=True`, cada imagem deve ser convertida para `float` e levada ao intervalo `[0, 1]` dentro de `__getitem__` (ou seja, em tempo real!).

1) Complete `__len__` e `__getitem__` para que `ImageDataset` envolva o `mnist` carregado acima. Em `__getitem__`, converta a imagem para `np.ndarray` com `dtype=float`.

2) Implemente a normalização min-max para `[0, 1]` aplicada **somente** quando `normalize=True` for passado no `__init__`:

````python
class ImageDataset:
    # TODO
    def __init__(self, base_dataset, normalize=False):
        # Construtor
        self.base_dataset = base_dataset
        self.normalize = normalize

    def __len__(self):
        # TODO Escreva sua solução aqui.
        return len(self.base_dataset)

    def __getitem__(self, idx):
        """Retorna (image: np.ndarray, label: int).

        - Converta a imagem do MNIST para np.ndarray float.
        - Se self.normalize, aplique a normalização min-max para [0, 1] (Exercício 3.1)
        """
        # TODO Escreva sua solução aqui.
        arrays = np.array(self.base_dataset[idx][0], dtype=np.float32)
        label = int(self.base_dataset[idx][1])

        if self.normalize == True:
            #passo 1: determinar menor e maior valor da matriz A
            minimo = arrays.min()
            maximo = arrays.max()

            #passo 2: normalizar
            C = (arrays - minimo)/(maximo - minimo)

            return C, label

        return arrays, label
````

### Código "boilerplate" para fazer o treinamento

````python
import numpy as np

class MNIST(torch.utils.data.Dataset):
    def __init__(self, normalize=False, train=False):
        super().__init__()
        from torchvision.datasets import MNIST

        self.base_dataset = MNIST(root='./data', train=train, download=True)
        self.normalize = normalize

    def __len__(self):
        return len(self.base_dataset)

    def __getitem__(self, idx):
        """Retorna (image: np.ndarray, label: int).

        - Converta a imagem do MNIST para np.ndarray float.
        - Se self.normalize, aplique a normalização min-max para [0, 1] (Exercício 3.1)
        """
        image, label = self.base_dataset[idx]
        image = np.array(image, dtype=np.float32)
        if self.normalize:
            image = self.min_max(image)

        image = torch.from_numpy(image).float()
        imagem_1d = image.flatten()

        return imagem_1d, label, image

    def min_max(self, array):
        array = np.array(array)
        return (array - array.min()) / (array.max() - array.min())
````

````python
import pytorch_lightning as pl
from torch.utils.data import DataLoader
from torchmetrics import Accuracy
````

````python
class DataModule(pl.LightningDataModule):
    def __init__(self):
        super().__init__()

    def setup(self, stage):
        # TODO aqui você deve inicializar seu dataset MNIST, lembre de passar o parâmetro de separar em conjunto de validação e treino!
        self.train = MNIST(train=True)
        self.val = MNIST(train=False)

    def train_dataloader(self):
        return DataLoader(self.train, batch_size=32, shuffle=True)

    def val_dataloader(self):
        return DataLoader(self.val, batch_size=32, shuffle=False)

````

````python
class Module(pl.LightningModule):

    def __init__(self, in_features=784, n_out=10):
        super().__init__()

        # Inicialize aqui sua camada linear.
        self.layer = MyLinearLayer(in_features, n_out, bias=True, optimize=True)

        # Para testar com a implementação oficial do PyTorch. TODO sua entrega não deve usar nn.Linear!
        #self.layer = torch.nn.Linear(in_features, n_out)

        self.acc = Accuracy(task="multiclass", num_classes=n_out)
        self.loss_fn = torch.nn.CrossEntropyLoss()

    def forward(self, x):
        return self.layer(x)

    def training_step(self, batch, batch_idx):
        x, y, z = batch
        # TODO como linearizar a imagem?

        logits = self(x)
        loss = self.loss_fn(logits, y)
        acc = self.acc(logits, y)

        self.log("train_loss", loss)
        self.log("train_acc", acc, prog_bar=True)
        return loss

    def validation_step(self, batch, batch_idx):
        x, y, z = batch

        # TODO como linearizar a imagem?

        logits = self(x)
        loss = self.loss_fn(logits, y)
        acc = self.acc(logits, y)

        self.log("val_loss", loss)
        self.log("val_acc", acc, prog_bar=True)


    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=1e-3)
````

````python
# Desafio opcional
import torch
import torchvision
import torchvision.transforms as transforms

class CIFAR(torch.utils.data.Dataset):
    def __init__(self, normalize=False, train=False):
        super().__init__()

        #transforma cifar10 em imagem na escala cinza
        self.base_dataset = torchvision.datasets.CIFAR10(root='./data', train=train, download=True, transform=transforms.Grayscale(num_output_channels=1))
        self.normalize = normalize

    def __len__(self):
        return len(self.base_dataset)

    def __getitem__(self, idx):
        """Retorna (image: np.ndarray, label: int).

        - Converta a imagem do MNIST para np.ndarray float.
        - Se self.normalize, aplique a normalização min-max para [0, 1] (Exercício 3.1)
        """
        image, label = self.base_dataset[idx]
        image = np.array(image, dtype=np.float32)
        if self.normalize:
            image = self.min_max(image)

        image = torch.from_numpy(image).float()
        imagem_1d = image.flatten()

        return imagem_1d, label, image

    def min_max(self, array):
        array = np.array(array)
        return (array - array.min()) / (array.max() - array.min())
````
